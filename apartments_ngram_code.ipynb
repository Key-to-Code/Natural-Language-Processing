```python
import math
import nltk
from nltk.util import ngrams
from collections import Counter

doc1 = "apartments are efficiently convenient"
doc2 = "apartments are cramped"
corpus = [doc1, doc2]

tokenized_corpus = [doc.split() for doc in corpus]

vocab = set(word for doc in tokenized_corpus for word in doc)
V = len(vocab)

def train_ngram_model(tokenized_corpus, n):
    counts = Counter()
    context_counts = Counter()
    for doc in tokenized_corpus:
        for ngram in ngrams(doc, n):
            counts[ngram] += 1
            if n > 1:
                context_counts[ngram[:-1]] += 1
            else:
                context_counts[()] += 1
    return counts, context_counts

def get_ngram_prob(ngram, counts, context_counts, V):
    n = len(ngram)
    context = ngram[:-1] if n > 1 else ()
    count = counts[ngram]
    context_count = context_counts[context]
    return (count + 1) / (context_count + V)

def calculate_sentence_probability(sentence, n, counts, context_counts, V):
    tokens = sentence.split()
    sentence_ngrams = list(ngrams(tokens, n))
    if not sentence_ngrams:
        return 0.0
    prob = 1.0
    for ngram in sentence_ngrams:
        p = get_ngram_prob(ngram, counts, context_counts, V)
        prob *= p
    return prob

for n in [1, 2, 3]:
    counts, context_counts = train_ngram_model(tokenized_corpus, n)
    p_pos = calculate_sentence_probability(doc1, n, counts, context_counts, V)
    p_neg = calculate_sentence_probability(doc2, n, counts, context_counts, V)
    print(f" {n}-Gram Model ")
    print(f"Positive Sentence Prob: {p_pos:.5f}")
    print(f"Negative Sentence Prob: {p_neg:.5f}")
    print()
```
