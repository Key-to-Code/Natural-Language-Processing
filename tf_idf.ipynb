```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "eating healthy salads with tomatoes and cucumber",
    "pasta with rich tomato sauce and cheese",
    "grilled chicken salad with olive oil dressing"
]

query = "healthy tomato salad"

all_texts = docs + [query]

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(all_texts) 
tfidf_array = tfidf_matrix.toarray() 

doc_vectors = tfidf_array[:len(docs)]
query_vector = tfidf_array[-1].reshape(1, -1)

cos_sim = cosine_similarity(doc_vectors, query_vector).flatten()

pca = PCA(n_components=2)
reduced_vectors = pca.fit_transform(tfidf_array) 

plt.figure(figsize=(8, 6))
labels = [f'D{i+1}' for i in range(len(docs))] + ['Query']
colors = ['blue', 'green', 'red', 'black'] 

for i, (x, y) in enumerate(reduced_vectors): 
    plt.scatter(x, y, color=colors[i % len(colors)], label=labels[i], s=100) 
    plt.text(x + 0.01, y + 0.01, labels[i], fontsize=12) 

plt.title('TF-IDF Vectors in 2D (PCA)') 
plt.xlabel('PCA Component 1') 
plt.ylabel('PCA Component 2') 
plt.legend() 
plt.grid(True) 

plt.savefig('food_tfidf_vectors.png')
plt.close() 

print("Cosine Similarities:")
for i, sim in enumerate(cos_sim, 1):
    print(f"D{i} vs Query: {sim:.3f}")
```
